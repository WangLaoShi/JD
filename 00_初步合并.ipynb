```python
# 列举所有的 sheet
import pandas as pd

# 读取Excel文件的所有sheet名称
file_path = '/Users/luosen/Documents/GitHub/JD/datas/src/招聘数据总表_2510.xlsx'
excel_file = pd.ExcelFile(file_path)

print("Excel文件中的所有sheet:")
print("=" * 50)
for i, sheet_name in enumerate(excel_file.sheet_names, 1):
    print(f"{i}. {sheet_name}")

print(f"\n总共有 {len(excel_file.sheet_names)} 个sheet")
```

```python
# 合并所有sheet到一个新的Excel文件中
import pandas as pd
import os

# 读取Excel文件的所有sheet
file_path = '/Users/luosen/Documents/GitHub/JD/datas/src/招聘数据总表_2510.xlsx'
excel_file = pd.ExcelFile(file_path)

print(f"开始合并 {len(excel_file.sheet_names)} 个sheet...")

# 创建一个列表来存储所有sheet的数据
all_dataframes = []

# 读取每个sheet并添加到列表中
for i, sheet_name in enumerate(excel_file.sheet_names):
    print(f"正在读取第 {i+1} 个sheet: {sheet_name}")
    df = pd.read_excel(file_path, sheet_name=sheet_name)
    
    # 添加一个列来标识数据来源的sheet
    df['数据来源'] = sheet_name
    
    all_dataframes.append(df)
    print(f"  - 读取了 {len(df)} 行数据")

# 合并所有数据框
print("\n正在合并所有数据...")
merged_df = pd.concat(all_dataframes, ignore_index=True)

print(f"合并完成！总共 {len(merged_df)} 行数据")

# 保存到新的Excel文件
output_file = '/Users/luosen/Documents/GitHub/JD/datas/src/招聘数据合并.xlsx'
print(f"\n正在保存到: {output_file}")

with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    merged_df.to_excel(writer, sheet_name='合并数据', index=False)

print("保存完成！")
print(f"文件已保存为: {output_file}")
print(f"合并后的数据包含 {len(merged_df)} 行，{len(merged_df.columns)} 列")
```

```python
# 查看合并后的数据概览
print("合并后的数据概览:")
print("=" * 50)
print(f"数据形状: {merged_df.shape}")
print(f"列名: {list(merged_df.columns)}")
print("\n前5行数据:")
print(merged_df.head())

print("\n数据来源统计:")
print(merged_df['数据来源'].value_counts())

print("\n数据类型:")
print(merged_df.dtypes)
```

```python
# 删除包含空值的数据并重新保存
import pandas as pd
import numpy as np

# 读取原始数据文件
input_file = '/Users/luosen/Documents/GitHub/JD/homeworks/招聘数据_整合.xlsx'
output_file = '/Users/luosen/Documents/GitHub/JD/homeworks/招聘数据_整合_去空.xlsx'

print("正在读取原始数据文件...")
df = pd.read_excel(input_file)

print(f"原始数据形状: {df.shape}")
print(f"原始数据包含 {df.shape[0]} 行，{df.shape[1]} 列")

# 检查空值情况
print("\n空值统计:")
print("=" * 50)
null_counts = df.isnull().sum()
print(null_counts[null_counts > 0])

# 显示空值比例
total_cells = df.shape[0] * df.shape[1]
total_nulls = df.isnull().sum().sum()
print(f"\n总空值数量: {total_nulls}")
print(f"空值比例: {total_nulls/total_cells*100:.2f}%")

# 删除包含空值的行
print("\n正在删除包含空值的行...")
df_cleaned = df.dropna()

print(f"清理后数据形状: {df_cleaned.shape}")
print(f"删除了 {df.shape[0] - df_cleaned.shape[0]} 行数据")
print(f"保留 {df_cleaned.shape[0]} 行数据")

# 保存清理后的数据
print(f"\n正在保存清理后的数据到: {output_file}")
df_cleaned.to_excel(output_file, index=False)

print("保存完成！")
print(f"文件已保存为: {output_file}")

# 验证保存的文件
print("\n验证保存的文件:")
df_verification = pd.read_excel(output_file)
print(f"验证文件形状: {df_verification.shape}")
print(f"验证文件是否还有空值: {df_verification.isnull().sum().sum()}")
```

```python
# 数据清理前后对比
print("数据清理前后对比:")
print("=" * 60)
print(f"原始数据: {df.shape[0]} 行")
print(f"清理后数据: {df_cleaned.shape[0]} 行")
print(f"删除的行数: {df.shape[0] - df_cleaned.shape[0]} 行")
print(f"数据保留率: {df_cleaned.shape[0]/df.shape[0]*100:.2f}%")

print("\n清理前后各列空值对比:")
print("-" * 60)
comparison = pd.DataFrame({
    '原始空值数': df.isnull().sum(),
    '清理后空值数': df_cleaned.isnull().sum()
})
comparison['空值减少数'] = comparison['原始空值数'] - comparison['清理后空值数']
print(comparison[comparison['原始空值数'] > 0])

print("\n清理后数据样本:")
print(df_cleaned.head())
```
